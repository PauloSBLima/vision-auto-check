<font color="yellow">Plano de ação para capturar o trabalho mais recente</font>

````python
%%bash
# 4.8 Carga geral na develop

# Garante que se QUALQUER comando falhar, o script para na hora
set -e

# 1. Salva e comita absolutamente tudo o que está modificado na pasta de trabalho atual
git add .
git commit -m "wip: salvando o estado mais recente e atualizado do Colab" || echo "Nenhuma alteração nova para comitar"

# 2. Descobre qual é a branch onde acabou de salvar esse trabalho
CURRENT_BRANCH=$(git branch --show-current)
echo "Trabalho recente capturado na branch: $CURRENT_BRANCH"

# 3. Vai para a develop e traz esse conteúdo atualizado para ela
git checkout develop
# Se estava trabalhando numa branch de feature, fazemos o merge dela na develop para atualizá-la
if [ "$CURRENT_BRANCH" != "develop" ]; then
    git merge "$CURRENT_BRANCH" -m "merge: atualiza develop com o trabalho mais recente de $CURRENT_BRANCH"
fi

# 4. Envia a develop atualizada para o GitHub
git push -u origin develop

# 5. Cria ou atualiza a 'main' com essa mesma base consolidada e atualizada
git checkout -b main 2>/dev/null || git checkout main
git push -u origin main --force

# 6. Retorna para a develop (ou para a sua feature) para você continuar o projeto
git checkout develop

echo "✓ Sucesso! O seu código mais recente do Colab agora está seguro na develop e na main!"
````

<font color="yellow">Plano de ação para os merges</font>

````python
%%bash
# 4.13 Merges - Integrações completas (Features -> Develop -> Main)
set -e

echo "=== INICIANDO FLUXO DE INTEGRAÇÃO DE MERGES ==="

# 1. Garante que a develop está limpa e atualizada
git checkout develop
git pull origin develop

# 2. Integra condicionalmente cada feature ativa na develop
for branch in feature/dataset-prep feature/fastapi-setup feature/yolo-finetuning; do
    if git show-ref --verify --quiet refs/remotes/origin/$branch || git show-ref --verify --quiet refs/heads/$branch; then
        echo "Integrando a branch: $branch"
        git checkout $branch
        git pull origin $branch || true
        git checkout develop
        git merge $branch -m "merge(develop): integra $branch de forma segura" || echo "Atenção: verifique eventuais conflitos em $branch"
    fi
done

# 3. Envia a develop consolidada para o GitHub
git push origin develop

# 4. Promove a develop atualizada para a branch main de produção
git checkout main
git pull origin main
git merge develop -m "release: promove develop consolidada para a main"
git push origin main

# 5. Retorna para a develop para manter o laboratório pronto para o próximo ciclo
git checkout develop

echo "✓ Ciclo completo de merges executado com sucesso! Projeto sincronizado no GitHub."
````
