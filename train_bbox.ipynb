## 🟢 3.0 Retomar Treinamento do Modelo

A melhor escolha para esta retomada é descartar os pesos do treinamento anterior e começar um novo treinamento do zero, utilizando os pesos pré-treinados padrão (no yolo11n.pt).

🔍 Por que começar do zero é a melhor estratégia aqui?
  1. Evitar <font color="orange">"Viés de Confusão"</font>: No treinamento anterior, o modelo foi exposto a uma mistura de abordagens (bounding boxes e polígonos de segmentação), o que gerou ruído nos mapas de características que ele aprendeu a extrair. Se dermos continuidade (resume) a partir daquele modelo, ele carregará esses aprendizados confusos.

  2. <font color="orange">Mudança Estrutural no Dataset</font>: Como o dataset_bbox agora é um ambiente puro focado estritamente em bounding boxes (com 58 imagens de treino, 16 de validação e 9 para testes), o ideal é que o modelo limpo aprenda esses padrões desde a primeira época, sem herdar os vícios da tentativa anterior.

  3. <font color="orange">Ponto de Partida Ideal</font>: Começar com os pesos base oficiais do YOLO combinados com o novo dataset limpo garante uma curva de aprendizado muito mais estável, limpa e previsível.

E sobre a quantidade de épocas ideal para iniciar com as 58 imagens disponíveis?

No treinamento inicial começamos com 50. Será que agora, já deveríamos iniciar em 100 épocas?

Análise: Aumentar para 100 épocas é uma boa ideia, mas há um fator importante a considerar por conta do tamanho do dataset: o volume de dados (58 imagens de treino).

Como o <font color="orange">dataset ainda é pequeno</font>, modelos YOLOv11n tendem a aprender e convergir muito rápido. Se treinados por muitas épocas sem dados suficientes, eles podem começar a sobreajustar (overfitting) — ou seja, memorizar perfeitamente as 58 imagens de treino e perder generalização.

Manter 50 épocas é excelente. Em datasets pequenos, o YOLO costuma estabilizar as curvas de perda (loss) bem antes disso. Além disso, estamos parametrizando o "patience", que é a parada antecipada.

````python
# Disparar o treinamento oficial do pipeline MVP para 'chave_combinada (inicialmente)'
resultados_treino = mf.treinar_modelo_yolo(
    caminho_yaml="dataset_bbox/data.yaml",
    modelo_base="yolo11n.pt",               # YOLOv11 nano (ideal para testes rápidos iniciais)
    epochs=50,                              # Quantidade segura para ver convergência inicial
    patience=15,                            # Para o treino se não houver melhora em 15 épocas seguidas
    imgsz=640,
    batch=16,
    nome_projeto="vision_auto_check_bbox",
    nome_experimento="chave_combinada_v2"
)
````

## 🧩 4.1 Análise do Treinamento Realizado

📊 O que os logs nos mostram?
  1. O pico ocorreu na Época 1: O modelo obteve seus melhores resultados logo na primeira época (mAP50 = 0.307), impulsionado pelos pesos pré-treinados do YOLO (yolo11n.pt).

  2. A Explosão de Erro (cls_loss): Entre as épocas 6 e 9, a perda de classificação (cls_loss) disparou violentamente para 15.66, fazendo as métricas caírem para quase zero.

  3. Parada Antecipada (Early Stopping): Como o modelo nunca mais conseguiu recuperar o desempenho da época 1 devido a essa desestabilização, o mecanismo de segurança entrou em ação e encerrou o treino na época 16.

🔍 Qual é a causa raiz provável? (O pulo do gato)

  - O que causou essa instabilidade e o salto na cls_loss por volta da época 6 é um fator muito comum ao treinar modelos leves (como o yolo11n) em datasets muito pequenos (58 imagens): o impacto agressivo das aumentações padrão, especialmente o Mosaic (mosaic=1.0).

🔍 O que está acontecendo?

  1. Poucos passos por época: Com 58 imagens e batch=16, o treino faz apenas 4 lotes (steps) por época. Isso significa que cada atualização de pesos tem um peso enorme no modelo.

  2. Efeito do <font color="orange">Mosaic</font>: O YOLO ativa por padrão o Mosaic, que junta 4 imagens em uma só para criar contexto. Em datasets pequenos, isso pode gerar sobreposições extremas ou distorções que confundem o aprendizado do modelo, causando picos repentinos de erro (loss spike) que o fazem perder o prumo.

🛠️ Como corrigir e estabilizar o próximo treino?

  - Para um dataset enxuto como este, o ideal é desligar ou reduzir drasticamente as aumentações pesadas nestas primeiras fases, permitindo que o modelo aprenda os padrões limpos das 58 imagens sem ruídos artificiais excessivos.

Vamos ajustar o código adicionando mosaic=0.0 para ver o modelo evoluir de forma linear e estável:

````python
# Disparar o treinamento ajustado para dataset pequeno (sem Mosaic agressivo)

from ultralytics import YOLO

# 1. Carrega o modelo base YOLOv11 nano oficial
model = YOLO("yolo11n.pt")

# 2. Dispara o treinamento com controle total dos parâmetros nativos
resultados_treino = model.train(
    data="dataset_bbox/data.yaml",
    epochs=50,
    patience=20,                                # Aumentamos levemente para dar margem
    imgsz=640,
    batch=16,
    mosaic=0.0,                                 # Desliga o mosaico para evitar distorções no dataset pequeno
    project="vision_auto_check_bbox",
    name="chave_combinada_v3"                   # Nova versão para acompanhar a mudança
)
````

Desta vez, a instabilidade sumiu! A perda de classificação (cls_loss) comportou-se de forma linear e previsível, caindo gradualmente de ~7.3 para ~2.1 ao longo das épocas. Isso prova que desligar o Mosaic (mosaic=0.0) removeu o ruído agressivo que estava quebrando o aprendizado do modelo nas primeiras tentativas.

🔍 O Diagnóstico Atual (Onde está o gargalo)
Apesar da estabilidade na perda, o modelo travou em um mAP50 baixo (~0.33) logo na época 2 e não evoluiu mais, acionando o Early Stopping na época 22.

O motivo para isso não é mais técnico do algoritmo, mas sim estrutural: 58 imagens de treino é um volume extremamente enxuto para treinar um detector de objetos do zero. O modelo consegue aprender o básico muito rápido (por isso o pico na época 2 usando os pesos pré-treinados), mas ele não tem massa de dados suficiente para generalizar padrões complexos e acertar o conjunto de validação.

## 4.3 Sobre a interrupção no treinamento para evitar o <font color="#32CD32">**Overfitting**</font>

Assim como no Keras, o treinamento com YOLO possui um mecanismo nativo de Early Stopping que monitora o desempenho do modelo na validação e interrompe o treino caso ele pare de melhorar (ou comece a sofrer overfitting).

O melhor de tudo é que ele <font color="yellow">**já vem ativado por padrão**</font> na biblioteca da Ultralytics!

Se repararmos no log aparece:

  - patience=100

Como funciona o Early Stopping no YOLO?

O parâmetro patience define o número de épocas consecutivas que o modelo vai aguardar sem apresentar melhora na métrica principal de validação (geralmente o mAP50) antes de cancelar o treinamento automaticamente.

Com a paciência configurada para 100, e o treino tinha um total de 50 épocas, ele roda até o final sem parar.

Se estivéssemos treinando por 300 épocas, por exemplo, e por volta da época 150 o modelo estagnasse por 100 épocas seguidas, o próprio YOLO pararia o treino sozinho e preservaria os melhores pesos salvos em best.pt.

<font color="orange">No caso deste pipeline, na função de treinamento definimos um parâmetro com paciência para 15 (patience=15). Será mais rigoroso e encerrará caso fique 15 épocas sem evoluir.</font>
